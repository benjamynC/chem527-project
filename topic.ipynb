## Question
Can RP-HPLC-UV be used to determine the percent of total venom proteom that is phospholipase A2 in spitting and non-spitting cobras?

primary dataset:  
https://figshare.com/articles/dataset/LC-UV_Chromatogram/19447583?file=34549724
Data accessed 11/07/2026   
secondary data:  
https://datacat.liverpool.ac.uk/1186/
Data accessed 11/07/2026  
https://doi.org/10.5524/102333
Data accessed 11/08/2026

```python
import pandas as pd

snakes = pd.DataFrame ({"full name": ["Hemachatus haemachatus", "Naja haje", "Naja nigricollis", "Naja naja", "Naja pallida"],
                        "abrev1": ["HEMHA", "NAJHA", "NAJMO", "NAJNA", "NAJPA"],
                        "abrev2": ["H_haemachatus", "N_haje", "N_nigricollis", "N_naja", "N_pallida"],
                        "abrev3": ["H. haemachatus", "N. haje", "N. nigricollis", "N. naja", "N. pallida"]
                        }).set_index("full name")


## LC-UV chromatogram data (primary data) is stored in the "Arrahman" folder, with one CSV file per species. 
## Each CSV file contains the following columns: retention time (min), A220 (unknown units), and A254 (unknown units).
print("LC-UV chromatogram data (primary data):")
for species in snakes.index:
    data = pd.read_csv("data/Arrahman/" + snakes.loc[species, "abrev1"] + ".csv")
    print(species, len(data), "rows")
    print(list(data.columns))
    print("Units: min, absorbance (unknown units), absorbance (unknown units)")
    print()

## top-down proteomics data (used for molar absorptivity weighting) is stored in the "Kazandjian" folder, with one .xlsx file per species.
## Each .xlsx file contains the following relevant columns in the "for alignment" tab: protein ID, abundance (%TIC), toxin family, toxin sub-type, and amino acid sequence.
## the H. haemachatus file has a "Alignment" tab instead of "for alignment", but the relevant columns are the same.
print("Top-down proteomics data (used for molar absorptivity weighting):")
h_data = pd.read_excel("data/Kazandjian/H_haemachatus_proteome_annotation.xlsx", sheet_name="Alignment")
print(snakes.index[0], len(h_data), "rows")
print(list(h_data.columns))
print("Units: % abundance (%TIC)")
print()

for species in snakes.index[1:]:
    data = pd.read_excel("data/Kazandjian/" + snakes.loc[species, "abrev2"] + "_proteome_annotation.xlsx", sheet_name="for alignment")
    print(species, len(data), "rows")
    print(list(data.columns))
    print("Units: % abundance (%TIC)")
    print()

## bottom-up proteomics data (used as a check) is stored in the "Nguyen" folder, with one .xls file for all species.
## NOTE the .xls file required an addiitonal package to read it (xlrd) 
## the .xls file contains rows for each toxin type and columns for each species with values reported as % abundance (mol/mol).
print("Bottom-up proteomics data (used as a check):")
check_data = pd.read_excel("data/Nguyen/TableS1-Supplementary Material revision.xls")
check_species = [name.strip() for name in check_data.columns[1:]]
print("one file for all, ", len(check_data), "rows")
print(list(check_data.columns))
print("Units: % abundance (mol/mol)")
for species in snakes.index:
    if snakes.loc[species, "abrev3"] in check_species:
        print(species, "present")
    else:
        print(species, "not present")
```
